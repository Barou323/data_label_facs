```python
import yt_dlp
import sys

def progress_hook(d):
    if d['status'] == 'downloading':
        percent = d.get('_percent_str', 'N/A')
        speed = d.get('_speed_str', 'N/A')
        eta = d.get('_eta_str', 'N/A')
        sys.stdout.write(f"\rDownload: {percent} | {speed} | ETA: {eta}")
        sys.stdout.flush()
    elif d['status'] == 'finished':
        sys.stdout.write("\nDownload 100%. Memproses file...\n")
        sys.stdout.flush()

def download_youtube(url):
    ydl_opts = {
        'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]/best',
        'outtmpl': '%(title)s.%(ext)s',
        'progress_hooks': [progress_hook],
        'quiet': True,
        'noprogress': True,
        'nocheckcertificate': True,
        'warnings': 'noWarnings'
    }

    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            ydl.download([url])
        print("SUKSES")
    except Exception as e:
        print(f"\nGAGAL: {e}")

url_video = "https://www.youtube.com/live/NZx392fNBD4"
download_youtube(url_video)
```
